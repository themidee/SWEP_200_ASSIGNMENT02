LIST COMPREHENSION: ONE OF THE METHODS THAT ENSURES OPTIMIZED MACHINE LEARNING TRAINING

```python
even_number_list = []
odd_number_list = [] # Initialize odd_number_list outside the loop...list is a class- it has attributes and methods, for example append
for i in MIDE:
  if i % 2 ==0:
    even_number_list.append(i)
  else:
    # No need to re-initialize here, it's done once before the loop
    odd_number_list.append(i)

#print(even_number_list)
#print(odd_number_list)

#LIST COMPREHENSION
even_number_list = [i for i in MIDE if i % 2 == 0]
odd_number_list = [i for i in MIDE if i % 2 != 0]

print(even_number_list)
print(odd_number_list)
```

```python
class Swep():
  def __init__(self, uni, level, dept, count = 47): #number of students = 47
    self.uni = uni
    self.level = level
    self.dept = dept
    self.count = count

  def product_of_student_count_with_length_of_uni(self): #method, belongs to class
    product = self.count * self.level
    return product

uni = "OAU"
level = 250
dept = "CHE"

#STEP1: INITIALIZE THE CLASS
swep = Swep(uni, level, dept)

#step2: PERFORM THE ACTION: to perform an operation under a clas, use the dot notation, for example initialie class.method)
swep.product_of_student_count_with_length_of_uni()
```
